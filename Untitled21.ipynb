## [1.Initial setup](#scrollTo=uOc-Ho1TkGjf&line=2)

```python
filepath = "https://raw.githubusercontent.com/tec03/Datasets/refs/heads/main/datasets/CreditCard.csv"
import pandas as pd
import numpy as np
df = pd.read_csv(filepath)
df.head()
```

```python
df.columns
```

```python
favs = ['CREDIT_LIMIT', 'PURCHASES', 'BALANCE']
```

```python
ndf = df[favs]
ndf.head()
```

This dataset gives the credit card usage behavior of customers of a bank. Each row gives the information of each customer over the last 6 months.

-CREDIT_LIMIT: This variable gives the maximum amount of credit available to the customer, it is a numerical variable

-PURCHASES: This variable gives the total amount spent by the customer on purchases, it is a numerical variable

-BALANCE: This variable gives the amount the customer currently owes on the card, it is a numerical variable

```python
ndf.shape
```

```python
ndf = ndf.dropna()
ndf.shape
```

##[2.RQ 1 Which credit limit group represents the largest share of the customers?](#scrollTo=uOc-Ho1TkGjf&line=2)


```python
ndf.head(2)
```

```python
def cus_cls(x):
  if 0<=x<=2000:
    return "0-2,000"
  elif 2000<x<=5000:
    return "2,001-5,000"
  elif 5000<x<=10000:
    return "5,001-10,000"
  return ">10,000"
```

```python
cus_cls(1500)
```

```python
ndf["Limitgp"] = ndf["CREDIT_LIMIT"].apply(cus_cls)
ndf.head()
```

```python
limit_gp_counts = pd.pivot_table(ndf,
    index="Limitgp",
    values="PURCHASES",
    aggfunc = "count")
limit_gp_counts
```

```python
new_names = {"PURCHASES": "count_LimitGP"}
limit_gp_counts.rename(columns=new_names,
    inplace=True)
limit_gp_counts.reset_index(inplace=True)
limit_gp_counts
```

```python
import matplotlib.pyplot as plt
explode_list=[0.01,0.03,0.03,0.02]
colors =["gold","green","coral","skyblue"]
limit_gp_counts["count_LimitGP"].plot(kind="pie",
    figsize=(15,6),
    autopct = "%1.1f%%",
    startangle=90,
    shadow=True,
    labels = None,
    pctdistance=1.1,
    explode=explode_list,
    colors=colors
    )
plt.axis("equal")
plt.legend(labels = limit_gp_counts.Limitgp,
    bbox_to_anchor =(1,0.5))
plt.title("Number of customers by credit limit group", y=1.1)
plt.show()
```

Insights

1. The two biggest slices are almost the same size. Customers with a limit of 0-2,000 (33.4%) and 2,001-5,000 (34.2%) make up about two thirds of the pie.

2. The smallest slice is the group above 10,000, with only 7.7% of customers. Very high credit limits are rare.

##[3.RQ 2 Which credit limit group uses the biggest share of its limit?](#scrollTo=uOc-Ho1TkGjf&line=2)

```python
ndf.shape
```

```python
ndf = ndf.dropna()
ndf.shape
```

```python
ndf["Limitgp"] = ndf["CREDIT_LIMIT"].apply(cus_cls)
```

```python
ndf["Usage"] = ndf["BALANCE"] / ndf["CREDIT_LIMIT"] * 100
ndf.head()
```

```python
usage_gp_mean = pd.pivot_table(ndf,
    index="Limitgp",
    values="Usage",
    aggfunc = "mean")
usage_gp_mean
```

```python
new_names = {"Usage": "mean_Usage"}
usage_gp_mean.rename(columns=new_names,
    inplace=True)
usage_gp_mean.reset_index(inplace=True)
usage_gp_mean
```

```python
import matplotlib.pyplot as plt
colors = ["gold", "green", "coral", "skyblue"]

plt.figure(figsize=(10,6))
for i, row in usage_gp_mean.iterrows():
    plt.bar(row["Limitgp"], row["mean_Usage"],
        color=colors[i],
        edgecolor="black",
        label=row["Limitgp"])
    plt.text(i, row["mean_Usage"] + 0.5, f"{row['mean_Usage']:.1f}%", ha="center")

plt.title("Average share of the credit limit used by group")
plt.xlabel("Credit limit group")
plt.ylabel("Average usage (%)")
plt.legend(title="Credit limit group")
plt.show()
```

Insights

1. Customers with the lowest credit limit (up to 2,000) use the biggest share of it, about 49% on average. They are the closest to reaching their limit.

2. The other three groups use a similar share, between 32% and 35%. Having a higher limit does not mean customers use more of it.
